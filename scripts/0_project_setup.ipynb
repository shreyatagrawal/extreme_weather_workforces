## Step 0: Mount Google Drive (Python runtime)

"Mounting" connects this notebook to your Google Drive so the code can read and write files there. Colab only mounts Drive while the runtime is **Python**.

1. Make sure the runtime is Python: **Runtime → Change runtime type → Python 3**.
2. Run the cell below.
3. In the pop-up, **pick the same Google account you're using in Colab and tick _Select all_**, then click Continue.
4. When the cell prints `Mounted at /content/drive`, you're connected.

**Backup: the Mount Drive button.** If the cell gives an error (for example `credential propagation was unsuccessful`) even after you tick *Select all*, mount from the sidebar instead, while you're still in Python:
1. Click the **folder icon** (Files) in the left sidebar.
2. At the top of the Files panel, click the **Mount Drive** button (the folder with the Google Drive logo).
3. You're connected when a `drive` folder (with `MyDrive` inside) appears in the Files panel.

````r
# Run this cell with a PYTHON runtime. Switch to R afterwards.
from google.colab import drive
drive.mount('/content/drive')
````

---
## Step 1: Create the project folder and set the working directory

**Switch the runtime to R now:** Runtime → Change runtime type → R.

Pick a short project name in `snake_case`: lowercase, words joined by underscores, no spaces or dashes. Use the same name for your GitHub repository later.

The **working directory** is R's "home base". Once it's set to the project folder, a path like `"raw_data/my_file.csv"` just works.

````r
# Name of the project folder (snake_case: no spaces, no dashes)
my_project_name <- "stream_temp_project"

# Build the full path. Your Google Drive lives at /content/drive/MyDrive
project_path <- file.path("/content/drive/MyDrive", my_project_name)

# Create the folder. showWarnings = FALSE: no warning if it already exists
dir.create(project_path, showWarnings = FALSE)

# Make the project folder R's working directory
setwd(project_path)

# Confirm: this should print a path ending in /stream_temp_project
getwd()
````

---
## Step 2: Create the subfolders

| Folder | What goes in it |
|--------|-----------------|
| `raw_data/` | Original data files, exactly as downloaded. **Never edited.** |
| `clean_data/` | Output of `1_data_clean.ipynb` |
| `figures/` | Plots and tables made by analysis scripts |
| `scripts/` | Every notebook, numbered in the order it runs |

````r
dir.create("raw_data",   showWarnings = FALSE)  # original data, never edited
dir.create("clean_data", showWarnings = FALSE)  # output of 1_data_clean
dir.create("figures",    showWarnings = FALSE)  # plots and tables from analysis scripts
dir.create("scripts",    showWarnings = FALSE)  # all notebooks, numbered in run order

# Check: you should see all four folders
list.files()
````
