```r
library(readr)
library(dplyr)
library(tidyr)
```

```r
raw_1996 <- read.csv("raw_data/StormEvents_details_1996.csv")
head(raw_1996)
```

```r
clean_1996 <- raw_1996 %>%
  rename_with(tolower) %>%
  filter_out(event_type == "Hail" & magnitude < 1) %>% #if event_type = hail, select only >=1 & if event_type is wind select >=50
  filter_out(event_type == "High Wind" & magnitude < 50) %>%
  filter_out(event_type == "Thunderstorm Wind" & magnitude < 50) %>%
  filter_out(event_type == "Strong Wind" & magnitude < 50) %>%
  filter_out(event_type == "Waterspout" & magnitude == 0)
```

```r
write.csv(clean_1996, "clean_data/clean_1996.csv")
list.files("clean_data")
```
